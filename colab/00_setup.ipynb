# Wasserstein Distributional Causal Forests: Colab setup

Run this notebook once in each Colab session. It clones the repository, installs the scientific Python dependencies, and verifies that the simulation runner is importable.

```python
from pathlib import Path
import os
import subprocess
import sys

REPO_URL = os.environ.get("WDCF_REPO_URL", "https://github.com/hugogobato/wasserstein-causal-forests.git")
REPO_DIR = Path("/content/wasserstein-causal-forests")
if not (REPO_DIR / "research/sim/runner.py").exists():
    subprocess.run(["git", "clone", REPO_URL, str(REPO_DIR)], check=True)

subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", str(REPO_DIR / "requirements-colab.txt")], check=True)
sys.path.insert(0, str(REPO_DIR / "research"))
os.chdir(REPO_DIR)
print("Repository:", REPO_DIR)
print("Python:", sys.version.split()[0])
```
